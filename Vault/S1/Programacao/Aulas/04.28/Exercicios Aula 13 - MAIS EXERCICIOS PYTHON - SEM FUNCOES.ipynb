### Exercício 2: Cálculo de Crescimento da Receita  
**Tópicos**: Laços, Listas, Cálculo percentual  

**Descrição**: A lista `receitas`, criada abaixo, contém a receita anual de uma empresa ao longo de vários anos.

Escreva um código que calcule a taxa de crescimento percentual da receita de um ano para o ano seguinte e imprima o crescimento de cada período.

A fórmula do crescimento percentual é:

$$
\text{crescimento} = \frac{\text{receita atual} - \text{receita anterior}}{\text{receita anterior}} \times 100
$$

Por exemplo, se as receitas forem `[100000, 110000, 115000]`, o código deve calcular:
- crescimento de 100000 para 110000;
- crescimento de 110000 para 115000.

Não é necessário criar função. Faça apenas a lógica principal do programa usando a lista `receitas`.

Sugestão:
- percorra a lista a partir da segunda posição;
- compare cada receita com a receita do ano anterior;
- calcule o crescimento percentual;
- imprima o resultado.


```python
from random import random
```

```python
receitas = [100_000.00]
for _ in range(10):
    receitas.append(round(receitas[-1]* (1 + random()/100), 2))
print(receitas)
```

### Exercício 3: Análise do Valor Médio dos Pedidos (AOV)  
**Tópicos**: Listas, Laços, Condicionais, Acumuladores  

**Descrição**: A lista `pedidos`, criada abaixo, contém os valores de diferentes pedidos de uma loja.

Escreva um código que calcule o Valor Médio dos Pedidos, também conhecido como AOV (*Average Order Value*), e conte quantos pedidos têm valor acima de R$ 1000,00.

O AOV é calculado por:

$$
\text{AOV} = \frac{\text{soma dos valores dos pedidos}}{\text{quantidade de pedidos}}
$$

Seu código deve:
- percorrer a lista `pedidos`;
- somar o valor total dos pedidos;
- contar quantos pedidos têm valor maior que 1000;
- calcular o valor médio dos pedidos;
- imprimir o AOV;
- imprimir a quantidade de pedidos acima de R$ 1000,00.

Não é necessário criar função. Faça apenas a lógica principal do programa usando a lista `pedidos`.


```python
from random import random
pedidos = [round(random() * 2000, 2) for _ in range(10)]
print(pedidos)
```
