1. Construa um programa em Python que, repetitivamente, pede ao usuário nomes de pessoas (somente o primeiro nome da pessoa e sem acentos). Essa repetição se encerra quando o usuário digitar um nome "vazio" (o usuário digita ENTER sem nenhum nome). Cada um dos nomes deve obrigatoriamente ser transformado em letras maiúsculas e ser colocado numa lista denominada LISTA (a string vazia não deve ser colocada na lista!). O programa deve imprimir então a lista LISTA.

2. Em seguida, o programa deve, a partir da lista LISTA, gerar outra lista, denominada LISTA2, contendo somente os nomes que começam por letra consoante (sugestão: crie uma outra lista contendo as vogais maiúsculas do nosso alfabeto). Finalmente, deve imprimir a lista LISTA2. 

Critério de correção (valor total = 3,0): 
- apenas imprimiu LISTA (não conseguiu imprimir LISTA2)  --> abate 1,5
- não tratou maiúscula/minúsculas --> abate 0,5
- a string vazia foi para a lista --> abate 0,5
- não há outros "meio-certo"
- programas que não possam ser executados não serão corrigidos.

### solução 1

````python
LISTA = []

while True:
    nome = input("Digite um nome (ENTER para sair): ")
    if nome == "":
        break
    LISTA.append(nome.upper())
print("LISTA:", LISTA)
vogais = ["A", "E", "I", "O", "U"]
LISTA2 = []
for nome in LISTA:
    if nome[0] not in vogais:
        LISTA2.append(nome)
print("LISTA2:", LISTA2)

LISTA3 =[]
for nome in LISTA:
    if nome[0] in vogais:
        LISTA3.append(nome)
print("LISTA3", LISTA3)

LISTA4 =[]
for nome in LISTA:
    if nome[-2] == "H":
        LISTA4.append(nome)
print("LISTA4", LISTA4)
````
