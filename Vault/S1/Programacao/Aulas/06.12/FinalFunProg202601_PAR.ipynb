## **Questão 2 (2,0 pontos)**

### ENUNCIADO:

Em um jogo, o usuário pode tentar acertar o sorteio de um número de 1 a 10.

```

numero = random.randint(1,10)
```

Após esse sorteio, o usuário tem 3 chances para acertar o número, sendo que a cada tentativa o usuário digita um número de 1 a 10.
* Se acertar na 1a tentativa, ganha 50 pontos e o jogo termina.
* Se acertar na 2a tentativa, ganha 20 pontos e o jogo termina.
* Se acertar na 3a tentativa, ganha  5 pontos e o jogo termina.
* Se não acertar, perde 15 pontos.

Crie um programa que mostre, no final, quantos pontos o usuário ganhou ou perdeu. Voce deve criar um laço (loop) para resolver essa questão. Não será aceito 3 inputs/if's em sequencia.

### CRITÉRIO DE CORREÇÃO (2,0):
> 1,0 ponto: estruturou as 3 tentativas com parada ao acertar <br>
> 0,5 ponto: definiu os pontos de cada tentativa <br>
> 0,5 ponto: apresentou corretamente o resultado <br>

Obs: Se definiu a pontuação certa mas não interrompe ao acertar perde a sub-rubrica da estrutura.

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 2 - NÃO DELETAR ESTA LINHA !!
import random
chances = 3
pontos = 0
numero = random.randint(1,10)
while chances > 0:
    chute = int(input(f"Qual seu Chute? você ainda tem este e mais {chances}!"))
    if chute == numero:
        if chances == 3:
            pontos += 50
            chances = 0
        elif chances == 2:
            pontos += 20
            chances = 0
        elif chances == 1: 
            pontos += 5
            chances = 0
    else:
        chances -=1
if chute != numero:
    pontos -= 15
else:
    pontos = pontos
print(f"Você terminou o jogo com: {pontos} pontos")
````

## **Questão 3 (2,0 pontos)**

### ENUNCIADO:

A empresa XPTO cobra **multa por atraso** no pagamento de fornecedores. O usuário informa o **valor da fatura**, os **dias de atraso** (inteiro) e o **tipo de contrato**.

A multa sobre o valor da fatura segue as faixas:

* 0 dias de atraso: sem multa
* de 1 a 10 dias: 2% do valor
* de 11 a 30 dias: 5% do valor
* acima de 30 dias: 10% do valor

Em seguida, o **tipo de contrato** ajusta a multa:
* MENSAL: multa cheia (ou seja, 100% da multa)
* ANUAL: multa pela **metade** (ou seja, 50% da multa)

Crie um programa que:\
a) pergunte 3 dados: o valor da fatura, os dias de atraso e o tipo de contrato;\
b) valide:
  - os dias de atraso devem ser um inteiro maior ou igual a zero (use `.isdigit()`)
  - o tipo de contrato deve ser `MENSAL` ou `ANUAL`, sendo que somente essas duas grafias devem ser aceitas.
  - Para o valor da fatura, suponha que o usuário sempre digita um número válido, ou seja, não é necessário verificação.
  - Se houver erro num dos dois campos (dias de atraso ou tipo de contrato), imprima `ENTRADA INVÁLIDA` e encerre.

c) caso não haja erro, imprima a multa e total no formato `Multa R$ X,XX` e `Total R$ X,XX`. Ou seja, números com apenas duas casas decimais e usando `,` como separador de decimal.

*Exemplo:*
fatura `R$1000, 20 dias, ANUAL → 5% = R$50, pela metade = Multa R$ 25,00`, `Total R$ 1025,00`.

**DICAS:**
* `"123".isdigit()` retorna `True`; `"12.3".isdigit()` e `"abc".isdigit()` retornam `False`.
* use `round()` para arredondar. Por exemplo: `round(1.234, 2)` retorna `1.23`.


### CRITÉRIO DE CORREÇÃO (2,0):
0,5: coletou os dados e validou (dias inteiro e tipo válido) corretamente.\
1,0: calculou a multa corretamente em todas as faixas, incluindo o caso de 0 dias e o ajuste ANUAL.\
0,5: apresentou a multa e o total a pagar no formato pedido.\
*Sem pontuação parcial dentro de cada rubrica.*

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 3 - NÃO DELETAR ESTA LINHA !!
valor_fatura = int(input("Valor da fatura?"))
dias_atraso = (input("Quantos dias de atraso?"))
tipo_contrato = input("Qual o tipo de contrato?")

if dias_atraso.isdigit():
    if tipo_contrato not in ["MENSAL", "ANUAL"]:
        print("ENTRADA INVÁLIDA")
    else:
        dias_atraso = int(dias_atraso)
        if dias_atraso == 0:
            multa = 0 * valor_fatura
        elif dias_atraso <= 10:
            multa = 0.02 * valor_fatura
        elif dias_atraso <= 30:
            multa = 0.05 * valor_fatura
        else:
            multa = 0.1 * valor_fatura
    if tipo_contrato == "MENSAL":
        multa = multa
    else:
        multa = multa / 2

    total = multa + valor_fatura
    multa_print = float(multa)
    total_print = float(total)
    print(f"Multa R${multa_print} e Total R${total_print}")
    
else:
    print("ENTRADA INVÁLIDA")
````

## **Questão 4 (2,0 pontos)**

### ENUNCIADO:

Uma plataforma de streaming deseja analisar as avaliações dos usuários para um conjunto de filmes.

Crie um programa que:

Solicite ao usuário quantos filmes serão avaliados. Não é necessário validar essa entrada, considere que o usuário vai digitar um número que pode ser convertido para `int` sem erro.

Para cada filme:
- Solicite o nome do filme.
- Solicite uma nota (número inteiro) entre 1 e 5.

Caso a nota esteja fora desse intervalo, peça novamente até que uma nota válida seja informada. Aqui também considere que o número digitado pode ser convertido para `int` sem erro.

Ao final, exiba:<br>
a) A média das notas.<br>
b) O(s) nome(s) do(s) filme(s) com a maior nota.<br>
c) Quantos filmes receberam nota igual ou superior a 4.<br>

### CRITÉRIO DE CORREÇÃO (2,0):
> 0,5 ponto: coletou quantidade, nomes e notas <br>
> 0,5 ponto: validou a nota (solicita novamente enquanto inválida) <br>
> 0,5 ponto: média correta <br>
> 0,5 ponto: identificou nome do(s) filme(s) top e a contagem de notas (itens b e c) <br>

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 4 - NÃO DELETAR ESTA LINHA !!

filmes = []
notas = []
num_filmes = int(input("Quantos filmes serão avaliados?"))
for i in range(num_filmes):
    nome_filme = (input("Nome do Filme?"))
    nota_filme = int(input("Nota filme?"))
    if nota_filme not in [1, 2, 3, 4, 5]:
        print("Erro, a nota deve ser entre 1 e 5.")
        nota_filme = int(input("Nota filme?"))
        notas.append(nota_filme)
    else:
        filmes.append(nome_filme)
        notas.append(nota_filme)       
c = 0
for nota in notas:
    if nota >= 4:
        c += 1
    else:
        c += 0
soma_notas = sum(notas)
media_notas = soma_notas / num_filmes
print(f"Media das notas: {media_notas}")
print(f"Total com nota maior que 4: {c}")
````
