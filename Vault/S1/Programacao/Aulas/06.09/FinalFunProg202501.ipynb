**Questão 3**

O valor do frete para os clientes da Dr. Paçoca depende do peso total do produto no pedido e da localização deste cliente.

* Se o peso for até 10.000g: R\$0,98 por quilo do produto
* Se o peso estiver entre 10.001g e 50.000g: R\$0,74 por quilo do produto
* Se o peso estiver entre 50.001g e 100.000g: R\$0,42 por quilo do produto
* Se o peso for acima de 100.000g: R$100 fixo, independente do peso


Se o estado for SP ou PR: sem adicional\
Se o estado for MG, RJ ou ES: dobra o valor do frete (apresentado acima)\
Não há entregas para outros estados

Você deve criar um programa que:\
a) receba do usuário o peso do pedido num valor inteiro em gramas e o destino (sigla da unidade federativa - estado ou distrito federal)\
b) valide as informações:\
    - o peso deve ser um número inteiro (não pode receber caractereres não numéricos, inclusive ponto decimal)\
    - destino deve receber apenas um dos cinco estados destacados acima. Além disso, se o usuário digitar "sp", o programa deverá reconhecer SP (em maiúsculas). \
    - Se houver erro, mostrar mensagem `ERRO DE ENTRADA` e não calcular o frete.\
c) caso não tenha erros, apresente o valor do frete.

\
CRITÉRIO DE CORREÇÃO (2,5):\
1,0: o programa coletou os dados do usuário e realizou as transformações necessárias\
1,5: o programa calcula corretamente os frete em todos os cenários e imprime o frete\

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 3 - NÃO DELETAR ESTA LINHA !!

peso_g = input("Peso do produto em gramas:")
estados = ["SP", "RJ", "MG", "PR", "ES"]
estados_com_adicional = ["MG", "RJ", "ES"]
if peso_g.isdigit():
    peso_g = int(peso_g)
    peso_kg = peso_g / 1000
    print(f"Peso computado com sucesso ({peso_kg}KG)")
    estado = input("Estado do pedido:")
    estado = estado.upper()
    if estado in estados:
        print(f"Estado computado com sucesso ({estado})")
        if peso_g <= 10000:
            frete = peso_kg * 0.98
        elif peso_g <= 50000:
            frete = peso_kg * 0.74
        elif peso_g <= 100000:
            frete = peso_kg * 0.42
        else:
            frete = 100
        if estado in estados_com_adicional:
            frete = frete*2
        else:
            frete = frete
        print(f"Seu custo de frete para enviar {peso_kg}KG ao estado de {estado} é R${frete}")
    else:
        print("ERRO DE ENTRADA")
    
else:
    print("ERRO DE ENTRADA")
````

# **Questão 4**

O arquivo indicado abaixo possui uma lista de pedidos do Mr. Paçoca. Porém, o preço unitário do código de produto 163C0150 está com o preço errado. Todos os produtos com este código devem ter preço unitário `R$0,29`.

Como isso acontece com uma certa frequência na empresa, crie um programa em que, após o carregamento do dataframe:\
a) solicite ao usuário o código do produto com erro e o preço correto (não é necessário verificação nessa etapa - suponha que ele vá inserir os dados corretamente)\
b) corrija para o valor correto de preço unitário em todas as linhas referentes ao código de produto inserido, utilizando obrigatoriamente as variáveis definidas em a)\
c) crie uma nova coluna com o total de cada pedido (preço unitário x quantidade)\
d) informe o total faturado global de todos os clientes

\
CRITÉRIO DE CORREÇÃO (2,5):\
0,5 ponto: o programa coleta os inputs (ítem a)\

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 4 - NÃO DELETAR ESTA LINHA !!

import pandas as pd
filename = 'MrPacoca.xlsx'
pedidos = pd.read_excel(filename)
pedidos.head()
````

````python
codigo_errado = input("Codigo do produto com valor errado:")
valor_correto = float(input("Valor correto:"))
pedidos.loc[pedidos["Código do Produto"] == codigo_errado, "Preço Unitário"] = valor_correto
pedidos["Total"] = pedidos["Preço Unitário"] * pedidos["Quantidade"]
faturamento_global = pedidos["Total"].sum()
print(f"O faturamento global foi de R${faturamento_global.round(2)}")
````

**Questão 5**

Cada pedido da Dr. Paçoca é organizado no formato de string, em que cada informação é separada por uma vírgula: Código do Cliente, Código do Produto, Quantidade.

A área de separação dos produtos recebe um conjunto de pedidos de uma vez só, em uma lista.

```
Exemplo:
pedidos = ["Cliente001,Produto002,250","Cliente007,Produto006,1120","Cliente002,Produto004,5400","Cliente005,Produto003,300"]

```

Escreva um programa que mostre a informação de cada um no seguinte formato:

\
**Código do Cliente - Código do Produto - Quantidade unidades**

\
O programa também deverá imprimir o número de pedidos e mostrar o total das quantidades.

No exemplo acima, o resultado é:
```
Cliente001 - Produto002 - 250 unidades
Cliente007 - Produto006 - 1120 unidades
Cliente002 - Produto004 - 5400 unidades
Cliente005 - Produto003 - 300 unidades

4 pedidos - 7070 unidades
```

CRITÉRIO DE CORREÇÃO (3,0):\
0,5 ponto: o programa faz uma varredura sobre os pedidos via um laço\
1,0 ponto: o programa separa as informações de cada pedido\
0,5 ponto: o programa imprime as informações de cada pedido no modelo solicitado\
1,0 ponto: o programa imprime uma unica linha totalizadora\
ATENÇÃO: O programa deverá funcionar para qualquer lista, não apenas para o exemplo informado.

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 5 - NÃO DELETAR ESTA LINHA !!

pedidos = ["Cliente001,Produto002,250","Cliente007,Produto006,1120","Cliente002,Produto004,5400","Cliente005,Produto003,300"]

acum = 0
total = 0
for i in pedidos:
    item = i.split(",")
    cliente = item[0]
    produto = item[1]
    unidade = item[2]
    acum += 1
    total += int(unidade)
    print(f"{cliente} - {produto} - {unidade} unidades")
print(f"{acum} pedidos - {total} unidades")
````
