## Exemplo 1

Escrever um programa que defina uma nota_final, e se:

* for maior ou igual a 6, apresente o status 'aprovado'
* se for menor do que 6, apresente o status 'não aprovado'

````python
nota_final = 6

if nota_final >= 6:
    status = 'aprovado'
else:
    status = 'não aprovado'

print('sua nota foi', nota_final, 'e seu status é', status)
````

## Exemplo 2

Escrever um programa que defina uma nota_final, e se:
* for maior ou igual a 6, apresente o status 'aprovado'
* se for maior ou igual a 5 e menor do que 6, apresente o status 'reaval'
* se for menor do que 5, apresente o status 'reprovado'

dica: ver https://www.geeksforgeeks.org/chaining-comparison-operators-python/

````python
nota_final = 4.9

if nota_final >= 6:
    status = 'aprovado'
elif nota_final >= 5 and nota_final < 6:
    status = 'reaval'
else:
    status = 'reprovado'

print('sua nota foi', nota_final, 'e seu status é', status)
````

````python
nota_final = 10

if nota_final >= 6:
    status = 'aprovado'
elif 5 <= nota_final < 6:
    status = 'reaval'
else:
    status = 'reprovado'

print('sua nota foi', nota_final, 'e seu status é', status)
````

## Exemplo 3

Escrever um programa que defina uma nota_final, e uma media_semestral, e se:
* for maior ou igual a 6, apresente o status 'aprovado'
* se for maior ou igual a 5 e menor do que 6, além de uma media_semestral maior ou igual a 6, apresente o status 'reaval'
* se for menor do que 5, apresente o status 'reprovado'

````python
nota_final = 9
media_semestral = 7

if nota_final >= 6:
    status = 'aprovado'
elif (5 <= nota_final < 6) and media_semestral >= 6:
    status = 'reaval'
else:
    status = 'reprovado'

print('sua nota foi', nota_final, 'e seu status é', status)
````
