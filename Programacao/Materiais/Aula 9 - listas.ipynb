```python
nome_completo = "Pedro de Alcântara João Carlos Leopoldo Salvador Bibiano Francisco \
                 Xavier de Paula Leocádio Miguel Gabriel Rafael Gonzaga"
nome_completo.split()
```

```python
# criar um string com as iniciais de cada nome, omitindo o 'de'
# por exemplo "Pedro de Alcantara" -> "PA"

iniciais = ""
for nome in nome_completo.split():
    if nome != "de":
        iniciais = iniciais + nome[0]
print(iniciais)
```
