Escrever um programa que:
* solicita ao usuario a digitação e entrada de um texto
* checa se o texto contém somente letras minusculas 
* exibir 'o texto contém somente minusculas', caso o ítem anterior seja positivo.
* exibir 'o texto contém minúsculas e maíusculas", caso negativo

<code> 
n = 543
centenas = n // 100
dezenas = ...
unidades = n - centenas * 100 - dezenas * 10
print(centenas)
print(dezenas)
print(unidades)
</code>
<br>

O código acima gera tres numeros que correspondem a centenas, dezenas e unidades. Por exemplo se n=543, o programa apresenta: 5, 4 e 3.<br>
Contudo, está faltando uma expressão para o cálculo das dezenas.<br>
Qual é a expressão correta?

Numa base de dados, o CPF está armazenado sem pontos, e sem hifen. Por exemplo, o CPF '123.456.789-00' está armazenado como 12345678900.

Escreva um programa que:
* solicita ao usuario a digitação e entrada de um CPF, considerando que pode fazê-lo digitando com '.' e '-'.
* o texto contendo o CPF deverá ser manipulado e comparado a um valor inteiro armazenado numa variável, p. ex. `cpf`. `cpf` é uma variável cujo tipo é `int`.
* se iguais, o programa deverá apresentar 'CPF identificado', e caso contrário, 'CPF não identificado' 

seguem dois CPF's para testarem seus programas:

<code> 
cpf = 71503115801
cpf = 8527444208 
</code>

Notar que o CPF pode começar com zero, mas um numero `int` não armazena zeros a esquerda.


Por questões de segurança, você não pode mostrar o número completo do CPF na tela.<br>
Assim, oculte os números deixando apenas os três primeiros e os dois últimos dígitos.<br>
Por exemplo, se o CPF for `261.667.140-92`, deverá ser exibido: `***.667.140-**`
