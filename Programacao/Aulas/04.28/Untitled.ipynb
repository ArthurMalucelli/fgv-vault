```python
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",  # ~1GB, roda em CPU tranquilo
    device_map="cpu"
)

out = pipe("Explica o que é WACC em 2 linhas:", max_new_tokens=100)
print(out[0]["generated_text"])
```
