### Q3 — taxa de entrega (2,5)

📌 *O que avaliar:* validar **antes** de converter (`isdigit` antes do `int()`); `.upper()` no período; faixas com `elif` e `<=`; dobro também sobre as taxas fixas (5,00 → 10,00 e 25,00 → 50,00 à noite).

**Saídas verificadas:**
- entrada `1500`, `d` → `Taxa de entrega: R$ 5.00`
- entrada `9000`, `N` → `Taxa de entrega: R$ 18.00`
- entrada `25000`, `D` → `Taxa de entrega: R$ 25.00`
- entrada `12.5`, `D` → `ERRO DE ENTRADA`
- entrada `3000`, `X` → `ERRO DE ENTRADA`

````python
dist = input("Distância em metros: ")
periodo = input("Período (D/N): ").upper()

if not dist.isdigit() or periodo not in ["D", "N"]:
    print("ERRO DE ENTRADA")
else:
    dist = int(dist)
    if dist <= 2000:
        taxa = 5.00
    elif dist <= 8000:
        taxa = (dist / 1000) * 1.20
    elif dist <= 15000:
        taxa = (dist / 1000) * 1.00
    else:
        taxa = 25.00
    if periodo == "N":
        taxa = taxa * 2
    print(f"Taxa de entrega: R$ {taxa:.2f}")
````

### Q4C — árvore de risco (2,0)

📌 *O que avaliar:* existência via `in pacientes["id"].values`; linha via `.iloc[0]` no filtro; regras na **ordem** com `elif` (a ordem importa: um obeso fumante com pressão alta é "consulta imediata", não "risco alto"); IMC com `:.1f`; travessão `–` copiado do enunciado.

**Saídas verificadas:**
- id `894` → `Paciente 894 (IMC: 26.3) – Risco: consulta imediata`
- id `741` → `Paciente 741 (IMC: 32.4) – Risco: risco alto`
- id `1` → `Paciente 1 (IMC: 31.2) – Risco: risco moderado`
- id `223` → `Paciente 223 (IMC: 28.0) – Risco: risco baixo`
- id `1500` → `Paciente não encontrado`

````python
import pandas as pd
pacientes = pd.read_csv("dataset_checkup_1000linhas.csv")

pid = int(input("id do paciente: "))

if pid not in pacientes["id"].values:
    print("Paciente não encontrado")
else:
    linha = pacientes[pacientes["id"] == pid].iloc[0]
    imc = linha["peso"] / (linha["altura"] ** 2)
    if linha["pressao_sistolica"] >= 140 or linha["pressao_diastolica"] >= 90:
        risco = "consulta imediata"
    elif imc >= 30 and linha["fumante"] == "sim":
        risco = "risco alto"
    elif imc >= 30 or linha["idade"] >= 60:
        risco = "risco moderado"
    else:
        risco = "risco baixo"
    print(f"Paciente {pid} (IMC: {imc:.1f}) – Risco: {risco}")
````

### Q5A — notas fiscais (3,0)

📌 *O que avaliar:* laço + `.split("|")`; `float()` antes de somar; contador (ou `len(registros)`); totalizadora com `:.2f`. Funciona pra qualquer lista.

**Saída verificada:**
```
Nota 98214 (LOJA03) - R$ 159.9
Nota 98215 (LOJA01) - R$ 89.5
Nota 98216 (LOJA12) - R$ 1230.0
Nota 98217 (LOJA03) - R$ 45.25

4 notas - Total: R$ 1524.65
```

````python
registros = ["LOJA03|98214|159.9", "LOJA01|98215|89.5",
             "LOJA12|98216|1230.0", "LOJA03|98217|45.25"]

contador = 0
total = 0
for reg in registros:
    partes = reg.split("|")
    loja = partes[0]
    nota = partes[1]
    valor = float(partes[2])
    contador += 1
    total += valor
    print(f"Nota {nota} ({loja}) - R$ {valor}")
print()
print(f"{contador} notas - Total: R$ {total:.2f}")
````

### Q5B — datas por extenso (3,0)

📌 *O que avaliar:* `split(" ")` pra separar data de hora; `len(partes)==2` distingue com/sem hora; `meses[int(mes)-1]` (lista base 0, mês base 1); singular/plural da hora e dos minutos; minutos 0 omitidos; montagem incremental da string.

**Saída verificada:**
```
25 de dezembro de 2025 à 1 hora e 30 minutos
1 de janeiro de 2026
15 de março de 2026 às 14 horas
7 de setembro de 2025 às 8 horas e 1 minuto
30 de junho de 2026 às 23 horas e 45 minutos
12 de outubro de 2025
```

````python
datas = ["25/12/2025 01:30", "01/01/2026", "15/03/2026 14:00",
         "07/09/2025 08:01", "30/06/2026 23:45", "12/10/2025"]

meses_do_ano = ['janeiro', 'fevereiro', 'março', 'abril', 'maio', 'junho',
                'julho', 'agosto', 'setembro', 'outubro', 'novembro', 'dezembro']

for reg in datas:
    partes = reg.split(" ")
    dia, mes, ano = partes[0].split("/")
    texto = f"{int(dia)} de {meses_do_ano[int(mes) - 1]} de {ano}"
    if len(partes) == 2:
        h, m = partes[1].split(":")
        h = int(h)
        m = int(m)
        if h == 1:
            texto += " à 1 hora"
        else:
            texto += f" às {h} horas"
        if m == 1:
            texto += " e 1 minuto"
        elif m > 1:
            texto += f" e {m} minutos"
    print(texto)
````
