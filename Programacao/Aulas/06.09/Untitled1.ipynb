```python
import pandas as pd
df = pd.DataFrame({
    "id":   [10, 25, 37],
    "nome": ["Ana", "Bruno", "Carla"]
})

print("--- 1. tabela inteira ---")
print(df)
print("--- 2. depois do filtro (repara: AINDA tem cabeçalho, é tabela) ---")
print(df[df["id"] == 25])
print("--- 3. depois do .iloc[0] (virou ficha deitada) ---")
linha = df[df["id"] == 25].iloc[0]
print(linha)
print("--- 4. lendo um campo da ficha ---")
print(linha["nome"])
```
