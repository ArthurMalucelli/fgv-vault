O arquivo carregado na célula abaixo possui a lista de pedidos de 2025. Os pedidos de um determinado **estado** sofrerão um reajuste de preço, e a empresa quer saber a diferença de faturamento.

Crie um programa em que: \
a) o pandas calcule o preço unitário de cada pedido (coluna Total / Quantidade) e o inclua em uma nova coluna chamada `Preço Unitário` \
b) o pandas some o valor de `reajuste` ao preço unitário **apenas** dos pedidos do estado informado (coluna Estado = estado) \
c) o pandas calcule o novo total de cada pedido (preço unitário reajustado × quantidade) \
d) seja calculada e apresentada a diferença entre o faturamento original e o faturamento após o reajuste

Utilize as variáveis `reajuste` e `estado` conforme código abaixo. \
O programa deve funcionar para **quaisquer** valores informados nestas variáveis.

**ATENÇÃO:** se você receber um alerta de `FutureWarning / ChainedAssignmentError`, ignore-o.

CRITÉRIO DE CORREÇÃO (2,5): \

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 4B - NÃO DELETAR ESTA LINHA !!

import pandas as pd
pedidos = pd.read_excel("pedidos.xlsx", index_col=0)

reajuste = 0.02
estado = "MG"
pedidos["Preço Unitário"] = pedidos["Total"] / pedidos["Quantidade"]
fat_antes = pedidos["Total"].sum()
pedidos.loc[pedidos["Estado"] == estado, "Preço Unitário"] += reajuste
pedidos["Total"] = pedidos["Preço Unitário"] * pedidos["Quantidade"]
fat_dps = pedidos['Total'].sum()
dif_fat = fat_dps - fat_antes
print(dif_fat)
````

Considere o arquivo `dataset_checkup_1000linhas.csv` (colunas: `id, idade, peso, altura, fumante, historico_familiar, diabetes, pressao_sistolica, pressao_diastolica`).

A classificação de **risco** de cada paciente segue as regras abaixo (nesta ordem de prioridade), onde IMC = peso / altura²:

1. Se `pressao_sistolica >= 140` **OU** `pressao_diastolica >= 90` → **consulta imediata**
2. Caso contrário, se `IMC >= 30` **E** `fumante == "sim"` → **risco alto**
3. Caso contrário, se `IMC >= 30` **OU** `idade >= 60` → **risco moderado**
4. Caso nenhuma anterior seja verdadeira → **risco baixo**

Monte um programa que: \
1. Carregue o dataset em um dataframe chamado `pacientes` (já feito na célula abaixo) \
2. Peça ao usuário um `id` (pode assumir que será um inteiro válido) \
3. Se o `id` **não existir**: apresente `Paciente não encontrado` \
4. Se existir: selecione a linha, calcule o IMC e apresente no formato **exato** (IMC com 1 casa decimal, inclusive o travessão `–`):

`Paciente 123 (IMC: 31.2) – Risco: risco alto`

* id `1` → `Paciente 1 (IMC: 31.2) – Risco: risco moderado`
* id `223` → `Paciente 223 (IMC: 28.0) – Risco: risco baixo`
* id `1500` → `Paciente não encontrado`

````python
# ESPAÇO PARA RESPOSTA DA QUESTAO 4C - NÃO DELETAR ESTA LINHA !!

import pandas as pd
pacientes = pd.read_csv("dataset_checkup_1000linhas.csv")
pacientes.head()
````

````python
input_id = int(input("ID:"))
if input_id in pacientes["id"].values:
    linha = pacientes[pacientes["id"] == input_id].iloc[0]
    ##atribui IMC ao paciente
    imc = linha["peso"] / linha["altura"] ** 2
    if linha["pressao_sistolica"] >= 140 or linha["pressao_diastolica"] >= 90:
        risco = "consulta imediata"
    elif imc >= 30 and linha["fumante"] == "sim":
        risco = "risco alto"
    elif imc >= 30 and linha["idade"] >= 60:
        risco = "risco moderado"
    else:
        risco = "risco baixo"
    print(f"Paciente {input_id} (IMC: {imc:.1f}) – Risco: {risco}")
else:
    print("Paciente não encontrado")
````
